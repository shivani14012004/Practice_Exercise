```python
# Problem Statement: Use map() to create a new list containing the square of each number.
a=[2,3,4,5]
list(map(lambda x:x**2,a))
```

```python
#Convert a list of numeric strings into integers using map().
a=['14','21','89']
list(map(int,a))
```

```python
#Problem Statement: Add corresponding elements of two lists using map().
a=[2,4,5]
b=[1,3,5]
list(map(lambda x,y:x+y,a,b))
```

```python
#Use filter() to select only even numbers from a list.
a=[1,2,3,4,5]
list(filter(lambda x:x%2==0,a))
```

```python
#Remove empty strings from a list using filter().
a=['om','','atharv','','shiva']
list(filter(lambda x:''.join(x),a))
```

```python
a=['om',' ','atharv',' ','shiva']
list(filter(lambda x:x!=' ',a))
```

```python
# Filter names whose length is greater than 5 characters.
a=['shivani','sakshi','om','swami']
list(filter(lambda x:len(x)>5,a))
```

```python
#Use reduce() to find the sum of all numbers in a list.
from functools import reduce
a=[1,2,3,4,5]
reduce(lambda x,y:x+y,a)
```

```python
#Use reduce() to multiply all numbers in a list.
a=[1,2,3,4,5]
reduce(lambda x,y:x*y,a)
```

```python
#Find the largest number from a list using reduce().
a=[10,20,30,85,97]
reduce(lambda x,y:x if x>y else y,a)
```

```python
#Combining filter() and map()
#Select even numbers from a list and then square them.
a=[1,2,3,4,5]
list(map(lambda x:x*x,filter(lambda x:x%2==0,a)))
```

```python
# Apply a normal function to every element of a list using map().
def add(price):
  return price + (price * 0.18)
prices = [100,200,300]
list(map(add,prices))
```

```python
#Problem Statement: Use reduce() with an initial value to add numbers.
from functools import reduce
a=[10,20,30]
reduce(lambda x,y:x+y,a,100)
```

```python
#Square Each Number Using map()
a=[1,2,3,4,5]
list(map(lambda x:x*x,a))
```

```python
# Convert a list of Celsius temperatures into Fahrenheit using map().
Celsius=[0,20,37,100]
list(map(lambda c:(c *9/5) +32,Celsius))
```

```python
#Given a list of student dictionaries, extract only the names using map().
student=[
    {'name':'Amit','marks':80},
    {'name':'shivani','marks':92},
    {'name':'shruti','marks':23}
]
list(map(lambda x:x['name'],student))
```

```python
#Write a Python program to filter only positive numbers from a list.
a=[1,2,3,-1,-2,-3]
list(filter(lambda x:x>0,a))
```

```python
#Filter students whose marks are greater than or equal to 40.
a=[
    {'name':'shivani','marks':98},
    { 'name':'sakshi','marks':76},
    {'name':'aarya','marks':34}
]
list(filter(lambda x:x['marks']>=40,a))
```

```python
# Filter all words that start with the letter P.
words=['python','java','pandas','ai']
list(filter(lambda x:x.startswith('p'),words))
```
