```python
#1. Even Numbers
numbers=[1,2,3,4,5]
tuple(filter(lambda x:x%2==0,numbers))
```

```python
#2. Odd Numbers
numbers=[1,2,3,4,5]
tuple(filter(lambda x:x%2!=0,numbers))
```

```python
#Positive Numbers
nums=[1,2,3,4,-1,-2,-3,-4]
tuple(filter(lambda x:x>0,nums))
```

```python
#4. Negative Numbers
nums=[1,2,3,4,-1,-2,-3,-4]
tuple(filter(lambda x:x<0,nums))
```

```python
#5. Numbers Greater Than 10
numbers=[1,2,3,4,5,10,20,30]
tuple(filter(lambda x:x>10,numbers))
```

```python
#Words Starting with 'P'
words=['python','pycharm','java','c++']
list(filter(lambda x:x.startswith('p'),words))
```

```python
#Remove Empty Strings
words = ["Python","","Java","","C"]
list(filter(lambda x:x!='',words))
```

```python
#Find Names Longer Than 4 Characters
words=['python','pycharm','java','c++']
list(filter(lambda x:len(x)>4,words))
```
