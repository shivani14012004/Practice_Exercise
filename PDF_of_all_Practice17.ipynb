```python
# Write a Python program to calculate factorial using recursion.
def factorial(n):
  if n==0:
    return 1
  return n*factorial(n-1)
print(factorial(5))
```

```python
# Create a function that accepts three numbers and returns the largest number.
def largest(a,b,c):
  if a>=b and a>=c:
    return a
  elif b>=c and b>=a:
    return b
  else:
    return c
print(largest(4,2,3))
```

```python
# Create a function that returns the factorial of a given number.
def factorial(num):
  fact=1
  while num >0:
    fact =fact*num
    num=num -1
  return fact
print(factorial(5))
```

```python
# Create a function that accepts a string and returns the reversed string.
def reverse(text):
  result=''
  for ch in text:
    result=ch+result
  return result
print(reverse('shivani'))
```

```python
# Create a function that counts vowels in a given string.
def count_vowel(text):
  vowel='aeiouAEIOU'
  count=0
  for i in text:
    if i in vowel:
      count=count+1
  return count
count_vowel('Shivani')


```

```python
# Create a function that checks whether a string is palindrome or not.
def palindrome(text):
  x=text.lower()
  return x==x[::-1]
palindrome('Madam')
```

```python
# Create a function that removes duplicate values from a list while maintaining order.
def duplicate(items):
  result=[]
  for i in items:
    if i not in result:
      result.append(i)
  return result
duplicate([1,2,2,3,3,4])
```

```python
# Create a function that validates a password. The password must have at least 8 characters and at least one digit.
def valid_password(password):
  if len(password) < 8:
    return False

  has_digit=False
  for ch in password:
    if ch.isdigit():
      has_digit= True
      break
  return has_digit
valid_password('python12344')
```
