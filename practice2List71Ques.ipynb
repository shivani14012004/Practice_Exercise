```python
#11. Write a Python function that takes two lists and returns True if they have at least one common member.
def common_member(num1,num2):
  for i in num1:
    if i in num2:
       return("True")
  return("False")
num1=[1,2,3,4]
num2=[3,4,5,6]
print(common_member(num1,num2))
```
