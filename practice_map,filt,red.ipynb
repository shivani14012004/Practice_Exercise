```python
#1. Understanding map() with Squaring Numbers
#Concept / Problem Statement: Use map() to create a new list containing the square of each number.
num=[1,2,3,4]
list(map(lambda x:x*x,num))
```

```python
#2. Using map() to Convert Strings into Integers
#Concept / Problem Statement: Convert a list of numeric strings into integers using map().
num=['10','20','30','40']
list(map(lambda x:int(x),num))
```

```python
#3. Using map() with Multiple Lists
#Concept / Problem Statement: Add corresponding elements of two lists using map().
num=[1,2,3,4]
num2=[1,3,2,5]
list(map(lambda x,y:x*y,num,num2))
```

```python
#4. Understanding filter() with Even Numbers
#Concept / Problem Statement: Use filter() to select only even numbers from a list.
num=[1,2,3,4]
list(filter(lambda x:x%2==0,num))
```

```python
#5. Using filter() to Remove Empty Strings
# Concept / Problem Statement: Remove empty strings from a list using filter().
words = ['Python', '', 'AI', '', 'Data']
list(filter(lambda x:x!='',words))
```

```python
#6. Using filter() with String Length
#Concept / Problem Statement: Filter names whose length is greater than 5 characters.
names = ['Amit', 'Pranav', 'Rahul', 'Shraddha', 'Yamin']
list(filter(lambda x:len(x)>5,names))
```

```python
#7. Understanding reduce() with Sum of Numbers
#Concept / Problem Statement: Use reduce() to find the sum of all numbers in a list.
from functools import reduce
num=[1,2,3,4]
reduce(lambda x,y:x+y,num)
```

```python
#8. Using reduce() to Find Product
#Concept / Problem Statement: Use reduce() to multiply all numbers in a list.
from functools import reduce
num=[1,2,3,4]
reduce(lambda x,y:x*y,num)
```

```python
#9. Using reduce() to Find Maximum Number
#Concept / Problem Statement: Find the largest number from a list using reduce().
num=[1,4,2,3,5]
max=reduce(lambda x,y:x if x>y else y,num)
print(max)
```

```python
#10. Combining filter() and map()
#Concept / Problem Statement: Select even numbers from a list and then square them.
num=[1,2,3,4,5]
list(map(lambda x:x*x ,filter(lambda x:x%2==0,num)))
```

```python
# 11. Using map() with User-Defined Function
# Concept / Problem Statement: Apply a normal function to every element of a list using map().
def add_def(price):
  return price+(price * 0.18)
price=[100,200,300]
final_price=list(map(add_def,price))
print(final_price)
```

```python
#12. Using reduce() with Initial Value
#Concept / Problem Statement: Use reduce() with an initial value to add numbers.
from functools import reduce
num=[10,20,30]
reduce(lambda x,y:x+y,num,100)
```

```python
#9. Count Total Characters in Words Using map() and reduce()
#Interview Question: Find the total number of characters in a list of words using map() and reduce().
from functools import reduce
words = ['AI', 'Python', 'Data']
len=list(map(lambda x:len(x),words))
reduce(lambda x,y:x+y,len)
```

```python
#10. Calculate Total Bill After Discount
#Interview Question: Use map() and reduce() to apply a 10% discount to prices and calculate the final total bill.
from functools import reduce
prices = [1000, 500, 800]
discount=list(map(lambda p:p-(p*0.10),prices))
reduce(lambda x,y:x+y,discount)

```

```python
#11. Filter Even Numbers and Find Their Sum
#Interview Question: From a list of numbers, filter even numbers and find their sum using filter() and reduce().
from functools import reduce
num=[1,2,3,4,5,6]
even=list(filter(lambda x:x%2==0,num))
reduce(lambda x,y:x+y,even)
```
