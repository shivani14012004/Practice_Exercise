```python
#Check Palindrome
#A palindrome reads the same forward and backward.eg:-madam
name=input('Enter a string:')

if name == name[::-1]:
  print('palindrome')
else:
  print('not palindrome')
```

```python
#4. Count Vowels
name=input('Enter a string:')
count=0
for ch in name:
  if ch in'aeiou':
    count=count+1
print('toal vowels:',count)
```

```python
name=input('Enter a string')
count=0
for ch in name.lower():
  if ch in 'aeiou':
    count=count+1
print('Total vowel',count)
```

```python
#Count Consonants

name=input('enter a string:')
count =0
for ch in name.lower():
  if ch.isalpha() and ch not in 'aeiou':
    count=count+1
print(count)
```

```python
#Count Uppercase and Lowercase
name=input('Enter a string:')
upper = 0
lower = 0
for ch in name:
  if ch.isupper():
    upper +=1
  elif ch.islower():
     lower +=1
print('uppercase:',upper)
print('lowercase:',lower)

```

```python
#7. Count Digits and Alphabets
name=input('Enter a string')
digit =0
alpha =0
for ch in name:
  if ch.isdigit():
    digit +=1
  elif ch.isalpha():
    alpha +=1
print('digit',digit)
print('alpha',alpha)
```
